```python
import json
import pandas as pd

with open('/content/train_separate_questions.json', 'r') as f:
    complex_data = json.load(f)

simple_rows = []

for contract in complex_data['data']:
    for paragraph in contract['paragraphs']:
        for qas in paragraph['qas']:

            if not qas.get('answers'):
                continue

            raw_id = qas['id']
            category = raw_id.split('__')[1].rsplit('_', 1)[0]

            for answer in qas['answers']:
                clause_text = answer['text'].strip()

                if len(clause_text) > 10:
                    simple_rows.append({
                        'text': clause_text,
                        'category': category
                    })

train_df = pd.DataFrame(simple_rows).drop_duplicates(subset=['text']).reset_index(drop=True)

print(train_df.tail())
```

```python
import json
import pandas as pd

with open('/content/test.json', 'r') as f:
    complex_data = json.load(f)

simple_rows = []

for contract in complex_data['data']:
    for paragraph in contract['paragraphs']:
        for qas in paragraph['qas']:

            if not qas.get('answers'):
                continue

            raw_id = qas['id']
            category = raw_id.split('__')[1].rsplit('_', 1)[0]

            for answer in qas['answers']:
                clause_text = answer['text'].strip()

                if len(clause_text) > 10:
                    simple_rows.append({
                        'text': clause_text,
                        'category': category
                    })

test_df = pd.DataFrame(simple_rows).drop_duplicates(subset=['text']).reset_index(drop=True)

print(test_df.tail())
```

```python
import pandas as pd

train_df['is_valid'] = False
test_df['is_valid'] = True

full_df = pd.concat([train_df, test_df], ignore_index=True)

print(f"Total rows: {len(full_df)}")
print(f"Training rows: {len(train_df)}")
print(f"Testing rows: {len(test_df)}")
```

```python
from fastai.text.all import *

dls = TextDataLoaders.from_df(
    full_df,
    text_col='text',
    label_col='category',
    valid_col='is_valid',
    bs=16
)

dls.show_batch(max_n=5)
```

```python
learn = text_classifier_learner(
    dls,
    AWD_LSTM,
    drop_mult=0.5,
    metrics=[accuracy]
)

learn.fine_tune(5, base_lr=2e-2)
```

```python
learn.export("legal_terms_conditions_model.pkl")
```

```python
from fastai.text.all import *

interp = ClassificationInterpretation.from_learner(learn)

interp.plot_confusion_matrix(figsize=(12, 12), dpi=100)
```
